### Exercise 1 — Sorting Arrays

1. Buat array 1D `temperatures` berisi `[30, 27, 33, 29, 31, 28]`.
2. Urutkan `temperatures` secara menaik dan menurun tanpa mengubah array asli.
3. Buat matriks `rainfall` berisi `[[8, 3, 6], [2, 9, 5], [7, 1, 4]]`.
4. Urutkan `rainfall` menggunakan `axis=0` dan `axis=1`, lalu jelaskan perbedaannya.

````python
# 1. Membuat array temperatures
temperatures = np.array([30, 27, 33, 29, 31, 28])

# 2. Mengurutkan secara menaik dan menurun
ascending = np.sort(temperatures)
descending = np.sort(temperatures)[::-1]

print("Array asli:", temperatures)
print("Ascending:", ascending)
print("Descending:", descending)

# 3. Membuat matriks rainfall
rainfall = np.array([
    [8, 3, 6],
    [2, 9, 5],
    [7, 1, 4]
])

# 4. Sorting berdasarkan axis
sort_axis0 = np.sort(rainfall, axis=0)
sort_axis1 = np.sort(rainfall, axis=1)

print("Rainfall:\n", rainfall)
print("Sorting axis 0:\n", sort_axis0)
print("Sorting axis 1:\n", sort_axis1)
````

### Exercise 2 — Filtering Arrays

Gunakan `scores = np.array([65, 80, 92, 74, 88, 95, 70])`.

1. Tampilkan elemen yang bernilai minimal 80 menggunakan Boolean mask.
2. Tampilkan nilai antara 70 dan 90 (termasuk keduanya).
3. Gunakan Fancy indexing untuk mengambil elemen pada indeks 0, 3, dan 5.
4. Gunakan `np.where()` untuk menemukan indeks nilai yang lebih besar dari 90.

````python
# Membuat array scores
scores = np.array([65, 80, 92, 74, 88, 95, 70])

# 1. Boolean mask: nilai minimal 80
mask_80 = scores >= 80
print("Boolean mask:", mask_80)
print("Nilai minimal 80:", scores[mask_80])

# 2. Nilai antara 70 dan 90
filter_range = scores[(scores >= 70) & (scores <= 90)]
print("Nilai 70 sampai 90:", filter_range)

# 3. Fancy indexing
selected = scores[[0, 3, 5]]
print("Fancy indexing:", selected)

# 4. np.where(): indeks nilai lebih dari 90
indices = np.where(scores > 90)
print("Indeks nilai > 90:", indices[0])
````

### Exercise 3 — Updating Arrays

Gunakan `inventory = np.array([[10, -1], [15, 8], [12, -3]])`.

1. Buat salinan dan ganti seluruh nilai negatif menjadi 0.
2. Tambahkan baris baru `[[20, 5]]` menggunakan `np.concatenate()`.
3. Tambahkan kolom `[[1], [2], [3], [4]]` pada hasil langkah 2.
4. Hapus baris pertama dan kolom terakhir menggunakan `np.delete()`; periksa `.shape` hasil akhir.

````python
# Membuat array inventory
inventory = np.array([
    [10, -1],
    [15, 8],
    [12, -3]
])

# 1. Membuat salinan dan mengganti nilai negatif
inventory_clean = inventory.copy()
inventory_clean[inventory_clean < 0] = 0

print("Inventory setelah diperbaiki:\n", inventory_clean)

# 2. Menambahkan baris baru
new_row = np.array([[20, 5]])
inventory_new = np.concatenate((inventory_clean, new_row), axis=0)

print("Setelah menambahkan baris:\n", inventory_new)

# 3. Menambahkan kolom baru
new_column = np.array([[1], [2], [3], [4]])
inventory_final = np.concatenate((inventory_new, new_column), axis=1)

print("Setelah menambahkan kolom:\n", inventory_final)

# 4. Menghapus baris pertama dan kolom terakhir
delete_row = np.delete(inventory_final, 0, axis=0)
delete_column = np.delete(delete_row, -1, axis=1)

print("Hasil akhir:\n", delete_column)
print("Shape:", delete_column.shape)
````

### Exercise 4 — Array Aggregation

1. Hitung total dan rata-rata seluruh pengamatan.
2. Hitung total setiap kolom (`axis=0`) dan rata-rata setiap baris (`axis=1`).
3. Hitung jumlah kumulatif pada setiap baris (`np.cumsum`).
4. Hitung rata-rata setiap baris menggunakan `keepdims=True`, lalu periksa bentuknya.

````python
# Membuat array visitors
visitors = np.array([
    [20, 25, 30],
    [18, 22, 28],
    [24, 27, 32]
])

# 1. Total dan rata-rata seluruh pengamatan
total = np.sum(visitors)
average = np.mean(visitors)

print("Total:", total)
print("Rata-rata:", average)

# 2. Total setiap kolom dan rata-rata setiap baris
column_total = np.sum(visitors, axis=0)
row_average = np.mean(visitors, axis=1)

print("Total setiap kolom:", column_total)
print("Rata-rata setiap baris:", row_average)

# 3. Jumlah kumulatif setiap baris
cumulative = np.cumsum(visitors, axis=1)

print("Cumulative sum:\n", cumulative)

# 4. Rata-rata setiap baris menggunakan keepdims
row_average_keep = np.mean(visitors, axis=1, keepdims=True)

print("Rata-rata dengan keepdims:\n", row_average_keep)
print("Shape:", row_average_keep.shape)
````

### Exercise 6 — Integrated Statistical Practice

Gunakan data nilai tiga ujian dari empat mahasiswa:

```python
student_results = np.array([[80, 75, 90],
                            [65, 70, 72],
                            [92, 88, 95],
                            [78, 85, 82]])
```

1. Hitung rata-rata nilai masing-masing mahasiswa.
2. Pilih baris mahasiswa yang rata-ratanya minimal 80.
3. Tentukan posisi mahasiswa dengan nilai rata-rata tertinggi memakai `np.argsort()`.
4. Tambahkan 5 poin pada setiap nilai, tetapi batasi nilai maksimum menjadi 100 dengan `np.where()`.
5. Hitung rata-rata per ujian pada data yang telah diperbarui.

````python
# Membuat data nilai mahasiswa
student_results = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [92, 88, 95],
    [78, 85, 82]
])

# 1. Rata-rata nilai masing-masing mahasiswa
student_average = np.mean(student_results, axis=1)

print("Rata-rata mahasiswa:", student_average)

# 2. Memilih mahasiswa dengan rata-rata minimal 80
selected_students = student_results[student_average >= 80]

print("Mahasiswa dengan rata-rata >= 80:\n", selected_students)

# 3. Posisi mahasiswa dengan rata-rata tertinggi
ranking = np.argsort(student_average)[::-1]

print("Urutan mahasiswa dari rata-rata tertinggi:", ranking)
print("Rata-rata tertinggi:", student_average[ranking[0]])

# 4. Menambahkan 5 poin, maksimum 100
updated_results = np.where(
    student_results + 5 > 100,
    100,
    student_results + 5
)

print("Nilai setelah penambahan:\n", updated_results)

# 5. Rata-rata setiap ujian setelah diperbarui
exam_average = np.mean(updated_results, axis=0)

print("Rata-rata setiap ujian:", exam_average)
````

### Exercise 6 — Integrated Statistical Practice

Gunakan data intensitas curah hujan (baris = hari, kolom = lokasi):

```python
rainfall = np.array([[12, 8, 5],
                     [10, 15, 7],
                     [20, 11, 9],
                     [8, 6, 4]])
```

1. Hitung rata-rata tiap lokasi, lalu lakukan **centering** menggunakan broadcasting.
2. Simpan hasil centering ke file `week11_centered_rainfall.npy`, kemudian muat kembali.
3. Ubah susunan data hasil pemuatan dari `(hari, lokasi)` menjadi `(lokasi, hari)` dengan `.T`.
4. Bagi data hasil transposisi menjadi dua bagian menurut kolom menggunakan `np.split()`.
5. Gabungkan kembali kedua bagian dengan `np.concatenate(axis=1)` dan buktikan hasilnya sama dengan data sebelum dibagi.

````python
# Membuat data curah hujan
rainfall = np.array([
    [12, 8, 5],
    [10, 15, 7],
    [20, 11, 9],
    [8, 6, 4]
])

print("Original Rainfall:\n", rainfall)

# 1. Menghitung rata-rata setiap lokasi
location_mean = np.mean(rainfall, axis=0)

print("Average per Location:", location_mean)

# Melakukan centering menggunakan broadcasting
centered_rainfall = rainfall - location_mean

print("Centered Rainfall:\n", centered_rainfall)

# 2. Menyimpan dan memuat hasil centering
np.save("week11_centered_rainfall.npy", centered_rainfall)

loaded_rainfall = np.load("week11_centered_rainfall.npy")

print("Loaded Rainfall:\n", loaded_rainfall)

# 3. Mengubah bentuk dari (hari, lokasi) menjadi (lokasi, hari)
transposed_rainfall = loaded_rainfall.T

print("Transposed Rainfall:\n", transposed_rainfall)
print("Shape:", transposed_rainfall.shape)

# 4. Membagi data menjadi dua bagian berdasarkan kolom
split_rainfall = np.split(transposed_rainfall, 2, axis=1)

part_1 = split_rainfall[0]
part_2 = split_rainfall[1]

print("Part 1:\n", part_1)
print("Part 2:\n", part_2)

# 5. Menggabungkan kembali kedua bagian
combined_rainfall = np.concatenate(
    (part_1, part_2),
    axis=1
)

print("Combined Rainfall:\n", combined_rainfall)

# Membuktikan kesamaan data
print("Arrays are equal:",
      np.array_equal(transposed_rainfall, combined_rainfall))
````
